```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('social_media_vs_productivity.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
print(df.isnull().sum())
```

```python

median_value = df['daily_social_media_time'].median()
df['daily_social_media_time'] = df['daily_social_media_time'].fillna(median_value)
```

```python
print(df.isnull().sum())
```

```python

mean_score = df['perceived_productivity_score'].mean()
df['perceived_productivity_score'] = df['perceived_productivity_score'].fillna(mean_score)
```

```python
print(df.isnull().sum())
```

```python
print(df['actual_productivity_score'].mean())
print(df['actual_productivity_score'].median())
```

```python
mean_val = df['actual_productivity_score'].mean()
df['actual_productivity_score'] = df['actual_productivity_score'].fillna(mean_val)

```

```python
print(df.isnull().sum())
```

```python
print(df['stress_level'].mean())
print(df['stress_level'].median())
print(df['stress_level'].mode())
```

```python
median_stress = df['stress_level'].median()
df['stress_level'] = df['stress_level'].fillna(median_stress)

```

```python
print(df.isnull().sum())
```

```python
print(df['sleep_hours'].min())
print(df['sleep_hours'].max())
```

```python
print(df['sleep_hours'].mean())
print(df['sleep_hours'].median())
print(df['sleep_hours'].mode())
```

```python
mean_sleep = df['sleep_hours'].mean()
df['sleep_hours'] = df['sleep_hours'].fillna(mean_sleep)
```

```python
print(df.isnull().sum())
```

```python
print(df['screen_time_before_sleep'].min())
print(df['screen_time_before_sleep'].max())
print('-----------------------------------')
print(df['screen_time_before_sleep'].mean())
print(df['screen_time_before_sleep'].median())
print(df['screen_time_before_sleep'].mode())
```

```python
mean_screen_time = df['screen_time_before_sleep'].mean()
df['screen_time_before_sleep'] = df['screen_time_before_sleep'].fillna(mean_screen_time)
```

```python
print(df.isnull().sum())
```

```python
print(df['job_satisfaction_score'].min())
print(df['job_satisfaction_score'].max())
print('-----------------------------------')
print(df['job_satisfaction_score'].mean())
print(df['job_satisfaction_score'].median())
print(df['job_satisfaction_score'].mode())
print('-----------------------------------')
print( (df['job_satisfaction_score'].isnull().sum() / len(df)) * 100,'%')
```

```python
mean_satisfaction = df['job_satisfaction_score'].mean()
df['job_satisfaction_score'] = df['job_satisfaction_score'].fillna(mean_satisfaction)
```

```python
print(df.isnull().sum())
```

```python
print(df.duplicated().sum())
```
