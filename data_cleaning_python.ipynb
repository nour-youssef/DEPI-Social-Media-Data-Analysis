```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('social_media_vs_productivity.csv')
```

```python
df.head()
```

```python
df["user_id"] = pd.Series(range(1,len(df)+1))
dim_gender = df[["gender"]].drop_duplicates().reset_index(drop=True)
dim_gender["gender_id"] = dim_gender.index +1
dim_gender = dim_gender[["gender_id","gender"]]
dim_gender.to_csv("dim_gender.csv",index=False)
dim_gender
```

```python
print(df.columns)
```

```python
df= df.merge(dim_gender,on="gender",how="left")
df=df.drop(columns=["gender"])
df.head(1)
```

```python
dim_job_type = df[["job_type"]].drop_duplicates().reset_index(drop=True)
dim_job_type["job_type_id"] = dim_job_type.index + 1
dim_job_type = dim_job_type[["job_type_id", "job_type"]]
dim_job_type.to_csv("dim_job_type.csv", index=False)
df= df.merge(dim_job_type,on="job_type",how="left")
df.drop(columns=["job_type"], inplace=True)
dim_job_type
```

```python
df.head(1)
```

```python
def categorize_age(age):
    if age < 20:
        return 'Youth (<20)'
    elif age < 30:
        return 'Young Adult (20-29)'
    elif age < 40:
        return 'Adult (30-39)'
    elif age < 50:
        return 'Middle-Aged (40-49)'
    else:
        return 'Senior (50+)'

df['age_group'] = df['age'].apply(categorize_age)

df
```

```python
dim_age_group = df[['age_group']].drop_duplicates().reset_index(drop=True)
dim_age_group['age_group_id'] = dim_age_group.index + 1
dim_age_group = dim_age_group[['age_group_id', 'age_group']]
dim_age_group.to_csv("dim_age_group.csv", index=False)
df = df.merge(dim_age_group, on='age_group', how='left')
df.drop(['age_group'], axis=1, inplace=True)
df.head(1)
```

```python
dim_social_platform = df[['social_platform_preference']].drop_duplicates().reset_index(drop=True)
dim_social_platform['social_platform_id'] = dim_social_platform.index + 1
dim_social_platform = dim_social_platform[['social_platform_id', 'social_platform_preference']]
dim_social_platform.to_csv("dim_social_platform.csv", index=False)
df = df.merge(dim_social_platform, on='social_platform_preference', how='left')
df.drop(['social_platform_preference'], axis=1, inplace=True)
df.head(1)
```

```python
def categorize_range(val):
    if val <= 50:
        return 'Low'
    elif val <= 70:
        return 'Medium'
    else:
        return 'High'

df['notifications_category'] = df['number_of_notifications'].apply(categorize_range)
dim_notifications = df[['notifications_category']].drop_duplicates().reset_index(drop=True)
dim_notifications['notifications_category_id'] = dim_notifications.index + 1
dim_notifications = dim_notifications[['notifications_category_id', 'notifications_category']]
dim_notifications.to_csv("dim_notifications_category.csv", index=False)

df = df.merge(dim_notifications, on='notifications_category', how='left')
df.drop(['notifications_category'], axis=1, inplace=True)
df

```

```python
median_stress = df['stress_level'].median()
df['stress_level'] = df['stress_level'].fillna(median_stress)
```

```python
print(df.isnull().sum())
```

```python
def categorize_stress(val):
    if val <= 3:
        return 'Low'
    elif val <= 7:
        return 'Medium'
    else:
        return 'High'

df['stress_category'] = df['stress_level'].apply(categorize_stress)
dim_stress = df[['stress_category']].drop_duplicates().reset_index(drop=True)
dim_stress['stress_category_id'] = dim_stress.index + 1
dim_stress = dim_stress[['stress_category_id', 'stress_category']]
dim_stress.to_csv("dim_stress_category.csv", index=False)
df = df.merge(dim_stress, on='stress_category', how='left')
df.drop(['stress_category'], axis=1, inplace=True)
df.head(1)
```
